```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Year": [2019, 2020, 2021, 2022, 2023, 2024, 2025],
    "Visitors": [39406181, 5896848, 7705943, 5700339, 28213003, 34928650, 40069360],
    "Gaming_Revenue": [292455, 60441, 86863, 42198, 183059, 226782, 247400]
}

df = pd.DataFrame(data)
df
```

```python
visitor_start = df.loc[0, "Visitors"]
gaming_start = df.loc[0, "Gaming_Revenue"]

df["Visitor_Index"] = df["Visitors"] / visitor_start * 100
df["Gaming_Index"] = df["Gaming_Revenue"] / gaming_start * 100

df.round(1)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(df["Year"], df["Visitor_Index"],
         marker="o", label="Visitor arrivals")
plt.plot(df["Year"], df["Gaming_Index"],
         marker="o", label="Gaming revenue")

plt.axhline(100, color="gray", linestyle="--",
            label="2019 level")

plt.title("Recovery of Macao Visitors and Gaming Revenue")
plt.xlabel("Year")
plt.ylabel("Recovery index  2019 equals 100")
plt.xticks(df["Year"])
plt.ylim(0, 110)
plt.grid(axis="y", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
result_2025 = df[df["Year"] == 2025][
    ["Year", "Visitor_Index", "Gaming_Index"]]

result_2025.round(1)
```
